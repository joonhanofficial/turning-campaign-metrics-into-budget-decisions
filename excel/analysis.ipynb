```python
# Check for missing values and duplicated campaign IDs

missing_values = df.isnull().sum()
duplicate_campaign_ids = df["campaign_id"].duplicated().sum()

print("Missing values by column:")
print(missing_values)

print("\nNumber of duplicated campaign IDs:")
print(duplicate_campaign_ids)
```

```python
# Create a focused campaign performance table

campaign_analysis = df[
    [
        "campaign_id",
        "campaign_objective",
        "platform",
        "ad_placement",
        "device_type",
        "creative_format",
        "retargeting_flag",
        "quarter",
        "impressions",
        "clicks",
        "conversions",
        "ad_spend",
        "revenue",
        "CTR",
        "CPC",
        "conversion_rate",
        "CPA",
        "ROAS",
        "profit"
    ]
].copy()

campaign_analysis.head()
```

```python
# Create benchmark thresholds for budget decision logic

roas_median = campaign_analysis["ROAS"].median()
cpa_median = campaign_analysis["CPA"].median()
conversion_rate_median = campaign_analysis["conversion_rate"].median()
profit_median = campaign_analysis["profit"].median()

print("ROAS median:", roas_median)
print("CPA median:", cpa_median)
print("Conversion rate median:", conversion_rate_median)
print("Profit median:", profit_median)
```

```python
# Create budget action recommendations

def assign_budget_action(row):
    if (
        row["ROAS"] > roas_median
        and row["CPA"] < cpa_median
        and row["profit"] > profit_median
        and row["conversions"] > 0
    ):
        return "Increase Budget"
    
    elif (
        row["ROAS"] < roas_median
        and row["profit"] < 0
    ):
        return "Reduce Budget"
    
    elif (
        row["conversions"] == 0
        or row["CPA"] > cpa_median
    ):
        return "Review"
    
    else:
        return "Hold"

campaign_analysis["budget_action"] = campaign_analysis.apply(assign_budget_action, axis=1)

campaign_analysis[
    [
        "campaign_id",
        "platform",
        "campaign_objective",
        "ad_spend",
        "revenue",
        "conversions",
        "CPA",
        "ROAS",
        "profit",
        "budget_action"
    ]
].head(10)
```

```python
# Summarise campaign performance by budget action

action_summary = (
    campaign_analysis
    .groupby("budget_action")
    .agg(
        campaigns=("campaign_id", "count"),
        total_spend=("ad_spend", "sum"),
        total_revenue=("revenue", "sum"),
        total_profit=("profit", "sum"),
        avg_roas=("ROAS", "mean"),
        avg_cpa=("CPA", "mean"),
        avg_conversion_rate=("conversion_rate", "mean")
    )
    .reset_index()
    .sort_values("total_profit", ascending=False)
)

action_summary
```

```python
# Summarise campaign performance by platform

platform_summary = (
    campaign_analysis
    .groupby("platform")
    .agg(
        campaigns=("campaign_id", "count"),
        total_spend=("ad_spend", "sum"),
        total_revenue=("revenue", "sum"),
        total_profit=("profit", "sum"),
        avg_roas=("ROAS", "mean"),
        avg_cpa=("CPA", "mean"),
        avg_conversion_rate=("conversion_rate", "mean"),
        increase_budget_campaigns=("budget_action", lambda x: (x == "Increase Budget").sum()),
        reduce_budget_campaigns=("budget_action", lambda x: (x == "Reduce Budget").sum()),
        review_campaigns=("budget_action", lambda x: (x == "Review").sum())
    )
    .reset_index()
    .sort_values("total_profit", ascending=False)
)

platform_summary
```

```python
# Create final campaign-level output for Excel

final_campaign_output = campaign_analysis[
    [
        "campaign_id",
        "platform",
        "campaign_objective",
        "ad_placement",
        "device_type",
        "creative_format",
        "retargeting_flag",
        "quarter",
        "impressions",
        "clicks",
        "conversions",
        "ad_spend",
        "revenue",
        "profit",
        "CTR",
        "CPC",
        "conversion_rate",
        "CPA",
        "ROAS",
        "budget_action"
    ]
].copy()

final_campaign_output = final_campaign_output.sort_values(
    by=["budget_action", "profit"],
    ascending=[True, False]
)

final_campaign_output.head(10)
```

```python
# Sort final campaign output by business priority

action_order = {
    "Increase Budget": 1,
    "Review": 2,
    "Hold": 3,
    "Reduce Budget": 4
}

final_campaign_output["action_priority"] = final_campaign_output["budget_action"].map(action_order)

final_campaign_output = final_campaign_output.sort_values(
    by=["action_priority", "profit"],
    ascending=[True, False]
)

final_campaign_output.head(10)
```

```python
# Remove helper column before export

final_campaign_output = final_campaign_output.drop(columns=["action_priority"])

final_campaign_output.head(10)
```

```python
# Export summary tables for Excel workbook

action_summary.to_csv(
    "../data/processed/budget_action_summary.csv",
    index=False
)

platform_summary.to_csv(
    "../data/processed/platform_performance_summary.csv",
    index=False
)

print("Export complete: budget_action_summary.csv")
print("Export complete: platform_performance_summary.csv")
```

```python
# Create platform budget action mix summary

platform_action_mix = (
    campaign_analysis
    .groupby(["platform", "budget_action"])
    .agg(
        campaigns=("campaign_id", "count"),
        total_spend=("ad_spend", "sum"),
        total_profit=("profit", "sum"),
        avg_roas=("ROAS", "mean"),
        avg_cpa=("CPA", "mean")
    )
    .reset_index()
)

platform_totals = (
    campaign_analysis
    .groupby("platform")
    .agg(total_platform_campaigns=("campaign_id", "count"))
    .reset_index()
)

platform_action_mix = platform_action_mix.merge(
    platform_totals,
    on="platform",
    how="left"
)

platform_action_mix["campaign_share"] = (
    platform_action_mix["campaigns"] / platform_action_mix["total_platform_campaigns"]
)

platform_action_mix = platform_action_mix.sort_values(
    by=["platform", "campaign_share"],
    ascending=[True, False]
)

platform_action_mix
```

```python
# Export platform budget action mix for Excel workbook

platform_action_mix.to_csv(
    "../data/processed/platform_budget_action_mix.csv",
    index=False
)

print("Export complete: platform_budget_action_mix.csv")
```
